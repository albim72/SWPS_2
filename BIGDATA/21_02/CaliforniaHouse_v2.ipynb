```python
import numpy as np
import pandas as pd
import joblib
import time
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
#wczytywanie danych
data = fetch_california_housing()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target #kolumna z cenami mieszkań
```

```python
#podział danych na zbiór treningowy i testowy
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['target']), df['target'], test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#inicjacja modeli
models = {
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=500,learning_rate=0.1, max_depth=4, random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=500, max_depth=10, random_state=42, n_jobs=-1)
}
```

```python
#trenowanie i ewaluacja
results = {}
for name, model in models.items():
  start_time = time.time()
  model.fit(X_train_scaled, y_train)
  train_time = time.time() - start_time

  y_pred = model.predict(X_test_scaled)
  mae = mean_absolute_error(y_test, y_pred) #średni błąd bezwzględny
  rmse = np.sqrt(mean_squared_error(y_test, y_pred)) #pierwiastek z błedu średniokwadratowego

  results[name] = {
      "Train time (s)": train_time,
      "MAE": mae,
      "RMSE": rmse
  }

  joblib.dump(model, f'{name}_model.pkl') #zapisywanie modelu do pliku
```

```python
#wyśweitlenie wyników
results_df = pd.DataFrame(results).T
results_df
```
